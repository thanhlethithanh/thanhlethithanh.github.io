```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Setup connection (Adjust 'your_db_name' to your actual database in SSMS)
engine = create_engine("mssql+pyodbc://localhost/PortfolioProject?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes")

# 2. Load the specific sheets from the Excel file
file_path = "data/Raw Data.xlsx"  # Path to where you moved the file
sheets = ['orders', 'customers', 'products']

for sheet in sheets:
    df = pd.read_excel(file_path, sheet_name=sheet)
    # Clean column names (remove spaces) for SQL compatibility
    df.columns = [c.replace(' ', '_') for c in df.columns]
    
    # Push to SQL Server
    df.to_sql(sheet, engine, if_exists='replace', index=False)
    print(f"Success: Sheet '{sheet}' is now a table in SQL Server.")
```
